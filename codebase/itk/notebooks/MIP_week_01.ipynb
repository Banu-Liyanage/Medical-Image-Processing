```python
from pathlib import Path
import itk
import matplotlib.pyplot as plt
import numpy as np

input_path = Path(r"C:\Users\Banuka\Documents\MIP\Medical-Image-Processing\codebase\itk\images\MIP_Image_01.dcm")

print("File exists:", input_path.exists())
print("Full path:", input_path.resolve())

image = itk.imread(str(input_path))

print("Image loaded successfully")
print("Image dimension:", image.GetImageDimension())
```

```python
image_array = itk.GetArrayViewFromImage(image)

print("Array shape:", image_array.shape)

if image_array.ndim == 2:
    display_image = image_array
elif image_array.ndim == 3:
    display_image = image_array[image_array.shape[0] // 2]
else:
    raise ValueError(f"Unsupported image shape: {image_array.shape}")

plt.imshow(display_image, cmap="gray")
plt.axis("off")
plt.show()
```
